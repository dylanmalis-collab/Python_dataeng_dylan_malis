# Mission 6

Ask the user for a string, convert it to a list of characters with the `list` function, then ask for a single character.

Remove the first occurrence of the character from the list, then display how many occurrences of the character remain in the list.

For instance if you enter `"Excellent work"` then the letter `"e"`, it will display:
```
There are still 2 copies of e in the list
```

````python
liste = list(str(input("Saississez une chaine de caractère : ")).lower())
lettre = str(input("Saisissez une lettre : ")).lower()
liste.remove(lettre)
copie = 0
for L in liste:
  if L == lettre:
    copie += 1
print(f"Il y a {copie} copies de {lettre} dans la liste.")
````

# Mission 7

Define three lists containing no duplicates. Then, indicate (however you like) the two that have the most elements in common (or all three if there's a tie).

For instance, with:
```
list_a = [1, 2, 4, 8, 16, 32]
list_b = [1, 2, 3, 5, 8, 13]
list_c = [2, 3, 5, 7, 11, 13]
```
It may display:
```
Lists B and C have the most common elements!
```

````python
list_a = [2, 3, 5, 6, 17, 33]
list_b = [2, 3, 4, 6, 9, 14]
list_c = [2, 3, 5, 7, 11, 13]
pointab = 0
pointac = 0
pointbc = 0
for loop in range(len(list_a)):
  if list_a[loop] == list_b[loop]:
    pointab += 1
    if list_a[loop] == list_c[loop]:
      pointac += 1
      pointbc += 1
  elif list_b[loop] == list_c[loop]:
      pointbc += 1
if pointab > pointac and pointab > pointbc:
  print("Les listes a et b on plus d'éléments en commun.")
elif pointac > pointbc:
  print("Les listes a et c on plus d'éléments en commun.")
elif pointbc > pointac:
  print("Les listes b et c on plus d'éléments en commun.")
elif pointab == pointac == pointbc:
  print("Les 3 listes ont autant d'éléments en commun.")
````

````python
#autre méthode
list_a = [2, 3, 5, 6, 17, 33]
list_b = [2, 3, 4, 6, 9, 14]
list_c = [2, 3, 5, 7, 11, 13]
pointab = sum(1 for i in list_a if i in list_b)
pointac = sum(1 for i in list_a if i in list_c)
pointbc = sum(1 for i in list_b if i in list_c)

max_point = max(pointab, pointac, pointbc)

if max_point == pointab:
  print("Les listes a et b on plus d'éléments en commun.")
elif max_point == pointac:
  print("Les listes a et c on plus d'éléments en commun.")
elif max_point == pointbc:
  print("Les listes b et c on plus d'éléments en commun.")
elif max_point == pointab == pointac == pointbc:
  print("Les 3 listes ont autant d'éléments en commun.")
````
