Below is a dictionary containing the number of French campuses that offer each kind of Wild Code School course.

wildcodeschool = {
   "data_analyst": 4,

   "dev_web": "13"
}

Find a way to only display the keys that have int values.

```python
wildcodeschool =	{
  3: 4,                         #j'ai crée un int dans les clés, il n'y en avait pas.
  "data_scientist": 2,
  "dev_web": "13"
}

for i, clé in enumerate(wildcodeschool): #Voilà une méthode pour afficher uniquement les clés en integer
  if type(clé) == int:
    print(clé)
```

# Mission 2

For the same dictionary, find a way to display all the values that have a key that contains the letter "e".

```python
wildcodeschool = { "data_analyst": 4, "data_scientist": 2, "dev_web": "13" }
lettre = "e"
for i, clé in enumerate(wildcodeschool): # je sépare les clé dans une variable clé
  if lettre in clé:                      # si la clé contien "e" alors affiche la valeur auquel cette clé renvoi
    print(wildcodeschool[clé])
```

# Mission 3

Within the same dictionary, change the type of the value of the "dev_web" key to be an int. Then add a new key-value pair which will be "total_campus" : 19, using a **dictionary method**.

**Change the value of "dev_web" to be and int :**

```python
wildcodeschool = { "data_analyst": 4, "data_scientist": 2, "dev_web": "13" }

wildcodeschool["dev_web"] = int(wildcodeschool["dev_web"]) # On transforme en int le str "13"
wildcodeschool.update({"total_campus" : 19})               # On ajoute une nouvelle pair clé/valeur
print(wildcodeschool)
```

**Add a new key-value pair which will be "total_campus" : 19, using a dictionary method**

```python

wildcodeschool.update({"total_campus" : 19})               # On ajoute une nouvelle pair clé/valeur
print(wildcodeschool)
```

# Mission 4
